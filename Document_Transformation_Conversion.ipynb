# Document Transformation
The following small scripts transform several files of several formats into files of other formats.
That means, the following file formats are transformed_
- markdown files (.md; e.g. used for documentations) into PDF format (.pdf)
- LaTeX documents (.tex) into MS Word documents (.docx)

General hint: Transformations / Conversions may take some time!

## MD to PDF
Since it is a requirement for the conversion of MD to PDF using the following code to have pdflatex installed, this is checked in the first cell, first. 

If it is no accessible, it needs to be installed:

🪟 Windows
1. Install [MiKTeX](https://miktex.org/download)
2. Optional: During installation: Activate the “install missing packages on-the-fly” option
3. Afterwards: Restart the computer (so that PATH is updated)

🐧 Linux (z. B. Ubuntu/Debian) \
```sudo apt install texlive-latex-base``` \

Or even more comprehensive (more packages, more save) \
```sudo apt install texlive-full``` \

🍏 macOS \
```brew install --cask mactex``` \

Caution: MacTeX is big (~4 GB)

````python
# Check whether pdflatex (a LaTeX version in general) is installed
import subprocess

try:
    result = subprocess.run(["pdflatex", "--version"], capture_output=True, text=True, check=True)
    print("✅ pdflatex gefunden!")
    print(result.stdout.split('\n')[0])  # Only first line of version is shown
except FileNotFoundError:
    print("❌ pdflatex ist NICHT installiert oder nicht im PATH.")
except subprocess.CalledProcessError as e:
    print("⚠️ pdflatex wurde gefunden, aber ein Fehler ist aufgetreten:", e)
````

````python
# MD to PDF
import subprocess
from pathlib import Path

# Input file (Markdown) & Output file (PDF)
md_file = Path("ODK_Implementation_Docs.md")  # Adjust
pdf_file = md_file.with_suffix(".pdf")

# Pandoc command
# --from markdown: explicit markdown input
# --to pdf: automatic PDF output
# --standalone: turns it into a complete HTML/PDF document
# --pdf-engine: engine like pdflatex, xelatex or tectonic
# -V linkcolor=blue: makes links blue and clickable

try:
    subprocess.run([
        "pandoc",
        str(md_file),
        "-o", str(pdf_file),
        "--from", "markdown",
        "--to", "pdf",
        "--standalone",
        "--pdf-engine=pdflatex",     # Alternative: xelatex or tectonic
        "-V", "linkcolor=blue"       # Clickable hyperlinks
    ], check=True)
    print(f"✅ PDF creation successful: {pdf_file}")
except subprocess.CalledProcessError as e:
    print("❌ Error during conversion:", e)
````

## LaTeX to DOCX
✅ Requirements:
1. MikTeX / LaTeX engine
   If not installed, yet: Install [MiKTeX](https://miktex.org/download)
2. Pandoc
   🪟 Windows
   Install [Pandoc for Windows](https://github.com/jgm/pandoc/releases/tag/3.7.0.2)
   Or using Chocolatey:
   ```choco install pandoc```

Caution: Pandoc cannot analyze or reconstruct PDF files, but reads .tex directly as text.

````python
import subprocess
from pathlib import Path

# Input file (.tex) and output file (.docx)
tex_file = Path("example.tex")
docx_file = tex_file.with_suffix(".docx")

try:
    subprocess.run([
        "pandoc",
        str(tex_file),
        "-o", str(docx_file),
        "--from=latex",
        "--to=docx"
    ], check=True)
    print(f"✅ Erfolgreich konvertiert: {docx_file}")
except subprocess.CalledProcessError as e:
    print("❌ Error during conversion:", e)
except FileNotFoundError:
    print("❌ Pandoc is not installed or not in PATH.")
````
